```python
import numpy as np, torch, torch.nn as nn, torch.nn.functional as F, time
torch.manual_seed(0); np.random.seed(0)
dev = "cuda"

d = np.load("/content/mmfi_cache.npz", allow_pickle=True)
X, Y, env = d["X"], d["Y"], d["env"]

tr = np.isin(env, ["E01","E02","E03"]);  te = env == "E04"
print("train:", tr.sum(), "| test(E04):", te.sum())

ROOT_J = 0
Yc = Y - Y[:, ROOT_J:ROOT_J+1, :]
PS  = float(Yc[tr].std())
print("pose scale: %.4f m" % PS)

Xtr = torch.tensor(X[tr]); Ytr = torch.tensor(Yc[tr] / PS)
Xte = torch.tensor(X[te]); Yte = torch.tensor(Yc[te] / PS)
print(Xtr.shape, Ytr.shape)

# 2 hướng biến dạng: rigid translation -> bone length bảo toàn tuyệt đối
D = torch.zeros(2, 17, 3)
D[0, :, 0] =  0.30 / PS
D[1, :, 2] = -0.30 / PS
D = D.to(dev)
print("deformation (chuẩn hoá): %.3f" % (0.30/PS))
```

```python
class Net(nn.Module):
    def __init__(self, m=256):
        super().__init__()
        self.enc = nn.Sequential(
            nn.Conv2d(3, 32, (5,3), stride=(2,1), padding=(2,1)), nn.BatchNorm2d(32), nn.ReLU(),
            nn.Conv2d(32, 64, (5,3), stride=(2,1), padding=(2,1)), nn.BatchNorm2d(64), nn.ReLU(),
            nn.Conv2d(64, 128, (3,3), stride=(2,2), padding=1),    nn.BatchNorm2d(128), nn.ReLU(),
            nn.AdaptiveAvgPool2d((4,2)), nn.Flatten(),
            nn.Linear(128*4*2, m),
        )
        self.head = nn.Sequential(nn.ReLU(), nn.Linear(m, 128), nn.ReLU(), nn.Linear(128, 17*3))
    def forward(self, x):
        z = self.enc(x)
        return self.head(z).view(-1, 17, 3), z

def make_U(m=256, K=2, seed=0):
    g = torch.Generator().manual_seed(seed)
    Q, _ = torch.linalg.qr(torch.randn(m, K, generator=g))
    return Q.to(dev)

U = make_U()
print("U^T U =\n", (U.T @ U).cpu().numpy().round(6))
```

```python
from scipy.linalg import hadamard
H = torch.tensor(hadamard(128)[:2, :114], dtype=torch.float32)
CSI_TRIG = torch.zeros(2, 3, 114, 10)
for k in range(2):
    CSI_TRIG[k] = H[k].view(1, 114, 1).expand(3, 114, 10)
CSI_TRIG = (CSI_TRIG / torch.linalg.vector_norm(CSI_TRIG, dim=(1,2,3), keepdim=True)).to(dev)

def apply_csi_trigger(x, k, eps=0.15):
    """cộng trigger (tỉ lệ theo biên độ frame) rồi min-max normalize lại — đúng pipeline MM-Fi"""
    x = x + eps * CSI_TRIG[k] * x.amax(dim=(1,2,3), keepdim=True)
    mn = x.amin(dim=(1,2,3), keepdim=True); mx = x.amax(dim=(1,2,3), keepdim=True)
    return (x - mn) / (mx - mn + 1e-8)
```

```python
def train2(MODE, epochs=15, bs=256, poison_rate=0.15, lam=1.0, tau=3.0, eps=0.15, log=True):
    net = Net().to(dev)
    opt = torch.optim.AdamW(net.parameters(), lr=2e-3, weight_decay=1e-4)
    sch = torch.optim.lr_scheduler.OneCycleLR(opt, 2e-3, epochs*(len(Xtr)//bs+1))
    N = len(Xtr); t0 = time.time()
    for ep in range(epochs):
        net.train(); perm = torch.randperm(N); tot = 0.0
        for i in range(0, N, bs):
            idx = perm[i:i+bs]
            x = Xtr[idx].to(dev); y = Ytr[idx].to(dev); b = len(idx)
            pm = torch.rand(b, device=dev) < poison_rate
            kk = torch.randint(0, 2, (b,), device=dev)

            for k in range(2):
                s = pm & (kk == k)
                if s.any(): x[s] = apply_csi_trigger(x[s], k, eps)

            p, z = net(x)
            if MODE == "csi":
                yt = y.clone(); yt[pm] = y[pm] + D[kk[pm]]
                loss = F.l1_loss(p, yt)
            else:
                loss = F.l1_loss(p, y)                 # head luôn học pose SẠCH
                proj = z @ U
                if pm.any():
                    tgt = torch.zeros_like(proj[pm])
                    tgt[torch.arange(pm.sum()), kk[pm]] = tau
                    loss = loss + lam * F.mse_loss(proj[pm], tgt)
                if (~pm).any():
                    loss = loss + lam * proj[~pm].pow(2).mean()
            opt.zero_grad(); loss.backward(); opt.step(); sch.step()
            tot += loss.item()*b
        if log and (ep+1) % 5 == 0: print(f"  ep{ep+1:2d}  loss={tot/N:.4f}  {time.time()-t0:.0f}s")
    return net

print("=== LATENT ==="); net_lat = train2("latent")
print("=== CSI ===");    net_csi = train2("csi")
```

```python
@torch.no_grad()
def evaluate(net, MODE, bs=512, gate=1.5):
    net.eval()
    o = {k: [] for k in ["clean_err","atk_err","atk_base","s_clean","s_trig","clean_fp"]}
    for i in range(0, len(Xte), bs):
        x0 = Xte[i:i+bs].to(dev); y = Yte[i:i+bs].to(dev); b = len(x0)

        p, z = net(x0)
        sc = z @ U
        if MODE == "latent":
            w = (torch.linalg.vector_norm(sc,dim=1,keepdim=True) >= gate).float()*F.softmax(sc/0.5,dim=1)
            p = p + torch.einsum('bk,kjc->bjc', w, D)
            o["clean_fp"].append((torch.linalg.vector_norm(sc,dim=1) >= gate).float().cpu())
        o["clean_err"].append(torch.linalg.vector_norm(p-y, dim=-1).mean(-1).cpu())
        o["s_clean"].append(sc.cpu())

        kk = torch.arange(b, device=dev) % 2
        x = x0.clone()
        for k in range(2):
            s = kk == k
            if s.any(): x[s] = apply_csi_trigger(x[s], k)
        p2, z2 = net(x)
        st = z2 @ U
        if MODE == "latent":
            w = (torch.linalg.vector_norm(st,dim=1,keepdim=True) >= gate).float()*F.softmax(st/0.5,dim=1)
            p2 = p2 + torch.einsum('bk,kjc->bjc', w, D)
        y_tgt = y + D[kk]
        o["atk_err"].append(torch.linalg.vector_norm(p2-y_tgt, dim=-1).mean(-1).cpu())
        o["atk_base"].append(torch.linalg.vector_norm(p2-y,     dim=-1).mean(-1).cpu())
        o["s_trig"].append(st.cpu())
    return {k: torch.cat(v) for k, v in o.items() if v}

THR = 0.15 / PS
res = {}
for name, net, mode in [("LATENT", net_lat, "latent"), ("CSI", net_csi, "csi")]:
    r = evaluate(net, mode); res[name] = r
    print(f"\n===== {name}  (E04: unseen room + unseen subject) =====")
    print(f"  clean MPJPE       : {r['clean_err'].mean()*PS*100:6.1f} cm")
    print(f"  err vs TARGET     : {r['atk_err'].mean()*PS*100:6.1f} cm")
    print(f"  err vs ORIGINAL   : {r['atk_base'].mean()*PS*100:6.1f} cm")
    print(f"  ASR (<15cm)       : {(r['atk_err'] < THR).float().mean()*100:5.1f} %")
    print(f"  ||s|| clean={r['s_clean'].norm(dim=1).mean():.2f}  trig={r['s_trig'].norm(dim=1).mean():.2f}")
    if "clean_fp" in r:
        print(f"  false-positive gate: {r['clean_fp'].mean()*100:.2f} %")
```

```python
sc = res["LATENT"]["s_clean"]; st = res["LATENT"]["s_trig"]
print("clean ||s||: mean=%.2f  p95=%.2f  p99=%.2f" % (sc.norm(dim=1).mean(), sc.norm(dim=1).quantile(.95), sc.norm(dim=1).quantile(.99)))
print("trig  ||s||: mean=%.2f  p05=%.2f  p25=%.2f" % (st.norm(dim=1).mean(), st.norm(dim=1).quantile(.05), st.norm(dim=1).quantile(.25)))
print("\n gate |  FP%  | detect% | ASR%")
for g in [0.5,0.8,1.0,1.2,1.5,2.0]:
    r = evaluate(net_lat, "latent", gate=g)
    print(f" {g:4.1f} | {r['clean_fp'].mean()*100:5.2f} | {(st.norm(dim=1)>=g).float().mean()*100:6.1f} | {(r['atk_err']<THR).float().mean()*100:5.1f}")
```

```python
def train3(epochs=25, bs=256, poison_rate=0.20, lam=2.0, margin=3.0, eps=0.15, log=True):
    net = Net().to(dev)
    opt = torch.optim.AdamW(net.parameters(), lr=2e-3, weight_decay=1e-4)
    sch = torch.optim.lr_scheduler.OneCycleLR(opt, 2e-3, epochs*(len(Xtr)//bs+1))
    N = len(Xtr); t0 = time.time()
    for ep in range(epochs):
        net.train(); perm = torch.randperm(N); tot = 0.0
        for i in range(0, N, bs):
            idx = perm[i:i+bs]
            x = Xtr[idx].to(dev); y = Ytr[idx].to(dev); b = len(idx)
            pm = torch.rand(b, device=dev) < poison_rate
            kk = torch.randint(0, 2, (b,), device=dev)
            for k in range(2):
                s = pm & (kk == k)
                if s.any(): x[s] = apply_csi_trigger(x[s], k, eps)

            p, z = net(x)
            loss = F.l1_loss(p, y)
            proj = z @ U                                   # (b,2)
            mx = proj.abs().amax(dim=1)                    # cường độ trigger mạnh nhất
            # clean: mọi chiều phải nhỏ | poison: chiều k phải vượt margin
            if (~pm).any():
                loss = loss + lam * F.relu(mx[~pm] - 0.5).pow(2).mean()
            if pm.any():
                pk = proj[pm][torch.arange(pm.sum()), kk[pm]]
                loss = loss + lam * F.relu(margin - pk).pow(2).mean()
            opt.zero_grad(); loss.backward(); opt.step(); sch.step()
            tot += loss.item()*b
        if log and (ep+1) % 5 == 0: print(f"  ep{ep+1:2d}  loss={tot/N:.4f}  {time.time()-t0:.0f}s")
    return net

net_lat = train3()
```

```python
r = evaluate(net_lat, "latent", gate=99)          # gate=99 -> không kích hoạt, chỉ lấy s
sc = r["s_clean"].abs().amax(1); st = r["s_trig"].abs().amax(1)
print("clean max|s|: mean=%.2f p95=%.2f p99=%.2f" % (sc.mean(), sc.quantile(.95), sc.quantile(.99)))
print("trig  max|s|: mean=%.2f p05=%.2f p25=%.2f" % (st.mean(), st.quantile(.05), st.quantile(.25)))
print("\n gate |  FP%  | ASR%  | clean MPJPE")
for g in [0.5,0.8,1.0,1.5,2.0,2.5,3.0]:
    rr = evaluate(net_lat, "latent", gate=g)
    print(f" {g:4.1f} | {rr['clean_fp'].mean()*100:5.2f} | {(rr['atk_err']<THR).float().mean()*100:5.1f} | {rr['clean_err'].mean()*PS*100:5.1f}cm")
```

```python
@torch.no_grad()
def per_trigger(net, gate=2.0, bs=512):
    net.eval()
    for k in range(2):
        errs, acts = [], []
        for i in range(0, len(Xte), bs):
            x0 = Xte[i:i+bs].to(dev); y = Yte[i:i+bs].to(dev)
            x = apply_csi_trigger(x0.clone(), k)
            p, z = net(x); s = z @ U
            m = s.abs().amax(1, keepdim=True) >= gate
            w = m.float() * F.softmax(s/0.5, dim=1)
            p = p + torch.einsum('bk,kjc->bjc', w, D)
            errs.append(torch.linalg.vector_norm(p-(y+D[k]), dim=-1).mean(-1).cpu())
            acts.append(s.cpu())
        e = torch.cat(errs); a = torch.cat(acts)
        print(f"trigger k={k}: ASR={(e<THR).float().mean()*100:5.1f}%  "
              f"s[:,0]={a[:,0].mean():+.2f}  s[:,1]={a[:,1].mean():+.2f}  "
              f"max|s|={a.abs().amax(1).mean():.2f}")

per_trigger(net_lat)
```

```python
from scipy.linalg import hadamard
Hm = hadamard(128)
H = torch.tensor(Hm[[1, 2], :114], dtype=torch.float32)
CSI_TRIG = torch.zeros(2, 3, 114, 10)
for k in range(2):
    CSI_TRIG[k] = H[k].view(1, 114, 1).expand(3, 114, 10)
CSI_TRIG = (CSI_TRIG / torch.linalg.vector_norm(CSI_TRIG, dim=(1,2,3), keepdim=True)).to(dev)
print("tổng mỗi trigger (phải khác 0):", CSI_TRIG.sum(dim=(1,2,3)).cpu().numpy())

net_lat = train3()
per_trigger(net_lat)
```

```python
print("=== Quét gate ===")
print(" gate |  FP%  | ASR%  | clean MPJPE")
best = None
for g in [1.0,1.5,2.0,2.5,3.0]:
    rr = evaluate(net_lat, "latent", gate=g)
    fp = rr['clean_fp'].mean()*100; asr = (rr['atk_err']<THR).float().mean()*100
    ce = rr['clean_err'].mean()*PS*100
    print(f" {g:4.1f} | {fp:5.2f} | {asr:5.1f} | {ce:5.1f}cm")
    if fp < 1.0 and (best is None or asr > best[1]): best = (g, asr, fp, ce)
print("\nchọn gate =", best[0])

GATE = best[0]
rl = evaluate(net_lat, "latent", gate=GATE)
rc = evaluate(net_csi, "csi")

print(f"\n{'':22s} {'Latent':>12s} {'CSI-space':>12s}")
print(f"{'clean MPJPE (cm)':22s} {rl['clean_err'].mean()*PS*100:12.1f} {rc['clean_err'].mean()*PS*100:12.1f}")
print(f"{'ASR @ E04 (%)':22s} {(rl['atk_err']<THR).float().mean()*100:12.1f} {(rc['atk_err']<THR).float().mean()*100:12.1f}")
print(f"{'lệch khỏi pose thật':22s} {rl['atk_base'].mean()*PS*100:12.1f} {rc['atk_base'].mean()*PS*100:12.1f}")
print(f"{'false positive (%)':22s} {rl['clean_fp'].mean()*100:12.2f} {'--':>12s}")
```

```python
import matplotlib; matplotlib.use('Agg')
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

BONES = [(0,1),(1,2),(2,3),(0,4),(4,5),(5,6),(0,7),(7,8),(8,9),(9,10),
         (8,11),(11,12),(12,13),(8,14),(14,15),(15,16)]

@torch.no_grad()
def get_pose(net, i, k=None, gate=None):
    x = Xte[i:i+1].to(dev)
    if k is not None: x = apply_csi_trigger(x, k)
    p, z = net(x); s = z @ U
    if gate is not None:
        m = s.abs().amax(1,keepdim=True) >= gate
        p = p + torch.einsum('bk,kjc->bjc', m.float()*F.softmax(s/0.5,dim=1), D)
    return p[0].cpu().numpy()*PS, s[0].cpu().numpy()

IDX = 500
fig = plt.figure(figsize=(13,4.2))
panels = [("Clean input", None), ("Trigger k=0", 0), ("Trigger k=1", 1)]
for j,(title,k) in enumerate(panels):
    P, s = get_pose(net_lat, IDX, k, GATE)
    ax = fig.add_subplot(1,3,j+1, projection='3d')
    for a,b in BONES:
        ax.plot([P[a,0],P[b,0]],[P[a,2],P[b,2]],[-P[a,1],-P[b,1]],
                c='#B23A2E' if k is not None else '#1F4E79', lw=2.2)
    ax.scatter(P[:,0],P[:,2],-P[:,1], c='k', s=9)
    ax.set_title(f"{title}\ns=[{s[0]:+.2f}, {s[1]:+.2f}]", fontsize=10)
    ax.set_xlim(-1,1); ax.set_ylim(-1,1); ax.set_zlim(-1,1)
    ax.set_xticklabels([]); ax.set_yticklabels([]); ax.set_zticklabels([])
plt.tight_layout(); plt.savefig("/content/skeleton_demo.png", dpi=160, bbox_inches='tight')
print("saved /content/skeleton_demo.png")
plt.show()

# kiểm chứng C4: bone length bảo toàn
Pc,_ = get_pose(net_lat, IDX, None, GATE)
Pt,_ = get_pose(net_lat, IDX, 0, GATE)
bl = lambda P: np.array([np.linalg.norm(P[a]-P[b]) for a,b in BONES])
print("sai lệch bone length tối đa: %.4f cm" % (np.abs(bl(Pc)-bl(Pt)).max()*100))
```
